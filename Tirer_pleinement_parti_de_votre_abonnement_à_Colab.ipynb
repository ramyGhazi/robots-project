```
import os
import numpy as np
import pandas as pd
import tensorflow as tf
from gensim.models import Word2Vec
from xgboost import XGBClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix

print("📥 1. CHARGEMENT DES MODÈLES DEPUIS LE DRIVE...")
CHEMIN_DRIVE = "/content/drive/MyDrive/Pipeline_HDFS_Models2/"

w2v_model = Word2Vec.load(CHEMIN_DRIVE + 'word2vec.model')
n2v_model = Word2Vec.load(CHEMIN_DRIVE + 'node2vec.model')
xgb_model = XGBClassifier()
xgb_model.load_model(CHEMIN_DRIVE + 'xgb_final.json')
lstm_model = tf.keras.models.load_model(CHEMIN_DRIVE + 'lstm_final.h5', compile=False)

print("✅ Modèles chargés avec succès.")

# =====================================================================
# 🛡️ LE SECRET DE LA PRODUCTION : LE VOCABULAIRE FIGÉ
# On récupère les événements EXACTS mémorisés lors de l'entraînement.
# Cela garantit que XGBoost recevra toujours ses 29 colonnes !
# =====================================================================
vocabulaire_exact = sorted(list(w2v_model.wv.key_to_index.keys()))
evt_to_idx = {evt: idx for idx, evt in enumerate(vocabulaire_exact)}

print(f"🔒 Vocabulaire verrouillé à {len(vocabulaire_exact)} événements (Format attendu par XGBoost).")

# ---------------------------------------------------------------------
# 📂 2. CHARGEMENT DE LA MATRICE DE SÉQUENCES
# ---------------------------------------------------------------------
# ATTENTION : Utilisez bien le fichier des séquences temporelles ici !
PATH_INPUT_CSV = "/content/Event_sequence_matrix.csv"

if not os.path.exists(PATH_INPUT_CSV):
    raise FileNotFoundError(f"❌ Fichier {PATH_INPUT_CSV} introuvable ! Ne mettez pas la matrice d'occurrences ici.")

df_matrix = pd.read_csv(PATH_INPUT_CSV)
col_label = [c for c in df_matrix.columns if c.lower() in ['label', 'target', 'class']][0]
cols_temporelles = [c for c in df_matrix.columns if c not in [col_label, 'BlockId']]

# Isolation du jeu de test (pour évaluer)
_, df_test = train_test_split(df_matrix, test_size=0.3, random_state=42, stratify=df_matrix[col_label])
y_test = df_test[col_label].values

# ---------------------------------------------------------------------
# ⚙️ 3. RECONSTRUCTION DES DONNÉES DE TEST SELON LE MOULE DU MODÈLE
# ---------------------------------------------------------------------
print("⚙️ Transformation des données de test...")

# A. Extraction des séquences pures
sequences_test = []
for _, row in df_test[cols_temporelles].iterrows():
    seq_pure = [str(evt) for evt in row if pd.notna(evt) and str(evt) not in ['0', '0.0', 'pad', 'PAD']]
    sequences_test.append(seq_pure)

# B. Génération Matrice XGBoost (GARANTIE À 29 COLONNES)
X_test_occurrence = np.zeros((len(sequences_test), len(vocabulaire_exact)))
for idx, seq in enumerate(sequences_test):
    for evt in seq:
        if evt in evt_to_idx:
            X_test_occurrence[idx, evt_to_idx[evt]] += 1

# C. Génération Tenseur BiLSTM
X_test_sequences = np.zeros((len(sequences_test), 50, 128))
for idx_s, seq in enumerate(sequences_test):
    for idx_e, evt in enumerate(seq[:50]):
        v_w2v = w2v_model.wv[evt] if evt in w2v_model.wv else np.zeros(64)
        v_n2v = n2v_model.wv[evt] if evt in n2v_model.wv else np.zeros(64)
        X_test_sequences[idx_s, idx_e, :] = np.concatenate([v_w2v, v_n2v])

# ---------------------------------------------------------------------
# 🔮 4. PRÉDICTIONS
# ---------------------------------------------------------------------
print("🔮 Calcul des prédictions (XGBoost + BiLSTM)...")

P_XGB = xgb_model.predict_proba(X_test_occurrence)[:, 1]
P_LSTM = lstm_model.predict(X_test_sequences, batch_size=512, verbose=0).flatten()

P_Final = (0.60 * P_LSTM) + (0.40 * P_XGB)
y_pred_final = (P_Final >= 0.05).astype(int)

# ---------------------------------------------------------------------
# 🏆 5. RÉSULTATS
# ---------------------------------------------------------------------
print("\n" + "="*60)
print("🏆 PERFORMANCES DES MODÈLES SAUVEGARDÉS")
print("="*60)
print(classification_report(y_test, y_pred_final, target_names=['Normal (0)', 'Anomalie (1)']))
print("Matrice de Confusion :")
print(confusion_matrix(y_test, y_pred_final))
```

```
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, Model, preprocessing, backend as K
from gensim.models import Word2Vec
import networkx as nx
from node2vec import Node2Vec
from xgboost import XGBClassifier
from sklearn.model_selection import train_test_split

# --- 1. FONCTION FOCAL LOSS ---
def focal_loss(gamma=2.0, alpha=0.25):
    def focal_loss_fixed(y_true, y_pred):
        y_true = tf.cast(y_true, tf.float32)
        epsilon = K.epsilon()
        y_pred = K.clip(y_pred, epsilon, 1.0 - epsilon)
        alpha_t = y_true * alpha + (K.ones_like(y_true) - y_true) * (1 - alpha)
        p_t = y_true * y_pred + (K.ones_like(y_true) - y_true) * (1 - y_pred)
        fl = - alpha_t * K.pow((K.ones_like(y_true) - p_t), gamma) * K.log(p_t)
        return K.mean(K.sum(fl, axis=-1))
    return focal_loss_fixed

# --- 2. CHARGEMENT ET NETTOYAGE ---
def load_and_clean_data():
    df_occ = pd.read_csv('/content/Event_occurrence_matrix.csv', engine='python', on_bad_lines='skip')
    df_traces = pd.read_csv('/content/Event_traces.csv', engine='python', on_bad_lines='skip')

    for df in [df_occ, df_traces]:
        df.columns = df.columns.str.strip().str.replace('\ufeff', '')

    df = df_occ.merge(df_traces[['BlockId', 'Features']], on='BlockId')

    y = df['Label'].apply(lambda x: 1 if str(x).lower() in ['anomaly', 'fail'] else 0).values

    df['Features_clean'] = df['Features'].astype(str).str.replace('[', '').str.replace(']', '').str.replace('"', '')

    cols_to_drop = ['BlockId', 'Label', 'Features', 'Features_clean', 'Type']
    X_num = df.drop(columns=[c for c in cols_to_drop if c in df.columns]).values

    return X_num, y, df['Features_clean'], df

X_num, y, features_text, df_full = load_and_clean_data()
sentences = [s.split(',') for s in features_text]

# --- 3. EMBEDDINGS (W2V + N2V) ---
w2v_model = Word2Vec(sentences, vector_size=64, window=5, min_count=1)
w2v_vectors = np.array([np.mean([w2v_model.wv[word] for word in s if word in w2v_model.wv] or [np.zeros(64)], axis=0) for s in sentences])

G = nx.Graph()
for seq in sentences:
    for i in range(len(seq) - 1):
        G.add_edge(seq[i], seq[i+1])
n2v = Node2Vec(G, dimensions=64, walk_length=10, num_walks=20, quiet=True)
model_n2v = n2v.fit(window=5, min_count=1)
n2v_vectors = np.array([np.mean([model_n2v.wv[n] for n in s if n in model_n2v.wv] or [np.zeros(64)], axis=0) for s in sentences])

# --- 4. BI-LSTM AVEC FOCAL LOSS ---
tokenizer = preprocessing.text.Tokenizer()
tokenizer.fit_on_texts(features_text)
X_seq = preprocessing.sequence.pad_sequences(tokenizer.texts_to_sequences(sentences), maxlen=50)

# Définition du modèle
input_seq = layers.Input(shape=(50,))
x = layers.Embedding(input_dim=len(tokenizer.word_index) + 1, output_dim=64)(input_seq)
x = layers.Bidirectional(layers.LSTM(64))(x)
outputs = layers.Dense(1, activation='sigmoid')(x)
model_lstm = Model(inputs=input_seq, outputs=outputs)

# Compilation avec Focal Loss
model_lstm.compile(optimizer='adam', loss=focal_loss(gamma=2.0, alpha=0.25))
# Entraînement pour apprendre les patterns d'anomalies
model_lstm.fit(X_seq, y, epochs=5, batch_size=64, verbose=1)

# Extraction des features (on prend la couche avant la sortie finale)
feature_extractor = Model(inputs=model_lstm.input, outputs=model_lstm.layers[-2].output)
seq_features = feature_extractor.predict(X_seq)

# --- 5. FUSION ET CLASSIFICATION AVEC SCALE_POS_WEIGHT ---
X_final = np.concatenate([seq_features, w2v_vectors, n2v_vectors, X_num], axis=1)
X_train, X_test, y_train, y_test = train_test_split(X_final, y, test_size=0.2, random_state=42)

# Calcul dynamique du poids pour XGBoost
ratio = sum(y == 0) / sum(y == 1)

meta_model = XGBClassifier(
    n_estimators=100,
    learning_rate=0.05,
    scale_pos_weight=ratio,
    eval_metric='logloss'
)
meta_model.fit(X_train, y_train)

print("Pipeline entraîné avec Focal Loss et Scale Pos Weight !")
```

```
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, Model, preprocessing, backend as K
from gensim.models import Word2Vec
import networkx as nx
from node2vec import Node2Vec
from xgboost import XGBClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix

# --- 1. FONCTION FOCAL LOSS ---
def focal_loss(gamma=2.0, alpha=0.25):
    def focal_loss_fixed(y_true, y_pred):
        y_true = tf.cast(y_true, tf.float32)
        epsilon = K.epsilon()
        y_pred = K.clip(y_pred, epsilon, 1.0 - epsilon)
        alpha_t = y_true * alpha + (K.ones_like(y_true) - y_true) * (1 - alpha)
        p_t = y_true * y_pred + (K.ones_like(y_true) - y_true) * (1 - y_pred)
        fl = - alpha_t * K.pow((K.ones_like(y_true) - p_t), gamma) * K.log(p_t)
        return K.mean(K.sum(fl, axis=-1))
    return focal_loss_fixed

# --- 2. CHARGEMENT ET SÉPARATION AU DÉBUT ---
def load_and_split_data():
    df_occ = pd.read_csv('/content/Event_occurrence_matrix.csv', engine='python', on_bad_lines='skip')
    df_traces = pd.read_csv('/content/Event_traces.csv', engine='python', on_bad_lines='skip')

    for df in [df_occ, df_traces]:
        df.columns = df.columns.str.strip().str.replace('\ufeff', '')

    df = df_occ.merge(df_traces[['BlockId', 'Features']], on='BlockId')
    df['Label_num'] = df['Label'].apply(lambda x: 1 if str(x).lower() in ['anomaly', 'fail'] else 0)
    df['Features_clean'] = df['Features'].astype(str).str.replace('[', '').str.replace(']', '').str.replace('"', '')

    # Séparation immédiate
    train_df, test_df = train_test_split(df, test_size=0.2, random_state=42, stratify=df['Label_num'])
    return train_df, test_df

train_df, test_df = load_and_split_data()

# --- 3. PRÉPARATION DES SÉQUENCES ---
sentences_train = [s.split(',') for s in train_df['Features_clean']]
sentences_test = [s.split(',') for s in test_df['Features_clean']]

tokenizer = preprocessing.text.Tokenizer()
tokenizer.fit_on_texts(train_df['Features_clean']) # Fit uniquement sur TRAIN

X_seq_train = preprocessing.sequence.pad_sequences(tokenizer.texts_to_sequences(train_df['Features_clean']), maxlen=50)
X_seq_test = preprocessing.sequence.pad_sequences(tokenizer.texts_to_sequences(test_df['Features_clean']), maxlen=50)

# --- 4. EMBEDDINGS (Entraînés sur Train seulement) ---
# Word2Vec
w2v_model = Word2Vec(sentences_train, vector_size=64, window=5, min_count=1)
def get_w2v_vectors(sentences, model):
    return np.array([np.mean([model.wv[word] for word in s if word in model.wv] or [np.zeros(64)], axis=0) for s in sentences])

w2v_vectors_train = get_w2v_vectors(sentences_train, w2v_model)
w2v_vectors_test = get_w2v_vectors(sentences_test, w2v_model)

# Node2Vec
G = nx.Graph()
for seq in sentences_train:
    for i in range(len(seq) - 1):
        G.add_edge(seq[i], seq[i+1])
n2v = Node2Vec(G, dimensions=64, walk_length=10, num_walks=20, quiet=True)
model_n2v = n2v.fit(window=5, min_count=1)
def get_n2v_vectors(sentences, model):
    return np.array([np.mean([model.wv[n] for n in s if n in model.wv] or [np.zeros(64)], axis=0) for s in sentences])

n2v_vectors_train = get_n2v_vectors(sentences_train, model_n2v)
n2v_vectors_test = get_n2v_vectors(sentences_test, model_n2v)

# --- 5. BI-LSTM (Entraîné sur Train) ---
input_seq = layers.Input(shape=(50,))
x = layers.Embedding(input_dim=len(tokenizer.word_index) + 1, output_dim=64)(input_seq)
x = layers.Bidirectional(layers.LSTM(64))(x)
outputs = layers.Dense(1, activation='sigmoid')(x)
model_lstm = Model(inputs=input_seq, outputs=outputs)

model_lstm.compile(optimizer='adam', loss=focal_loss(gamma=2.0, alpha=0.25))
model_lstm.fit(X_seq_train, train_df['Label_num'].values, epochs=5, batch_size=64, verbose=1)

feature_extractor = Model(inputs=model_lstm.input, outputs=model_lstm.layers[-2].output)
seq_features_train = feature_extractor.predict(X_seq_train)
seq_features_test = feature_extractor.predict(X_seq_test)

# --- 6. FUSION ET CLASSIFICATION ---
cols_to_drop = ['BlockId', 'Label', 'Features', 'Features_clean', 'Type', 'Label_num']
X_num_train = train_df.drop(columns=[c for c in cols_to_drop if c in train_df.columns]).values
X_num_test = test_df.drop(columns=[c for c in cols_to_drop if c in test_df.columns]).values

X_train_final = np.concatenate([seq_features_train, w2v_vectors_train, n2v_vectors_train, X_num_train], axis=1)
X_test_final = np.concatenate([seq_features_test, w2v_vectors_test, n2v_vectors_test, X_num_test], axis=1)

ratio = sum(train_df['Label_num'] == 0) / sum(train_df['Label_num'] == 1)

meta_model = XGBClassifier(n_estimators=100, learning_rate=0.05, scale_pos_weight=ratio, eval_metric='logloss')
meta_model.fit(X_train_final, train_df['Label_num'].values)

# --- 7. ÉVALUATION HONNÊTE ---
y_pred = meta_model.predict(X_test_final)
print(classification_report(test_df['Label_num'], y_pred))
```

```
# =============================================================================
#  PIPELINE HYBRIDE CORRIGÉ — Bi-LSTM (Word2Vec + Node2Vec) + XGBoost
#  Détection d'anomalies dans les logs HDFS
#
#  CORRECTIONS APPLIQUÉES :
#  [FIX 1] Sélection explicite des colonnes E → plus de risque Label_num dans XGBoost
#  [FIX 2] Node2Vec implémenté sur graphe construit depuis train uniquement
#  [FIX 3] Embedding hybride 128D (Word2Vec 64D + Node2Vec 64D) injecté dans Bi-LSTM
#  [FIX 4] Focal Loss implémentée (γ=2.0, α=0.75) pour déséquilibre de classes
#  [FIX 5] scale_pos_weight calculé dynamiquement pour XGBoost
#  [FIX 6] Fusion pondérée Weighted Soft Voting (0.6 × LSTM + 0.4 × XGB)
#  [FIX 7] Seuil décisionnel optimisé t=0.10 sur courbe précision-rappel
#  [FIX 8] Noms de features explicites dans SHAP (E1…E29 + LSTM_0…LSTM_127)
# =============================================================================

import os
import gc
import joblib
import numpy as np
import pandas as pd
import tensorflow as tf
import networkx as nx

from tensorflow.keras import layers, Model, backend as K
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.callbacks import EarlyStopping

from gensim.models import Word2Vec
from node2vec import Node2Vec

from xgboost import XGBClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import (classification_report, confusion_matrix,
                             precision_recall_curve, f1_score)
import shap
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

# =============================================================================
#  CONFIGURATION
# =============================================================================
CHECKPOINT_DIR = '/content/checkpoints'
os.makedirs(CHECKPOINT_DIR, exist_ok=True)

MAX_LEN       = 50     # longueur max des séquences Bi-LSTM
W2V_DIM       = 64     # dimension Word2Vec
N2V_DIM       = 64     # dimension Node2Vec
EMBED_DIM     = W2V_DIM + N2V_DIM  # 128D embedding hybride final
LSTM_UNITS_1  = 128    # unités première couche Bi-LSTM
LSTM_UNITS_2  = 64     # unités seconde couche Bi-LSTM
DROPOUT_RATE  = 0.3
BATCH_SIZE    = 64
EPOCHS        = 10
RANDOM_STATE  = 42

WEIGHT_LSTM   = 0.6    # poids fusion Bi-LSTM
WEIGHT_XGB    = 0.4    # poids fusion XGBoost
THRESHOLD     = 0.10   # seuil décisionnel optimisé


# =============================================================================
#  UTILITAIRES
# =============================================================================
def save_ckpt(obj, name):
    joblib.dump(obj, os.path.join(CHECKPOINT_DIR, name))

def load_ckpt(name):
    p = os.path.join(CHECKPOINT_DIR, name)
    return joblib.load(p) if os.path.exists(p) else None

def ckpt_exists(name):
    return os.path.exists(os.path.join(CHECKPOINT_DIR, name))


# =============================================================================
#  1. CHARGEMENT ET SPLIT
# =============================================================================
def load_and_split():
    """
    Charge les données et effectue le split AVANT tout calcul.
    [FIX 1] Label_num est créé ici mais sera exclu explicitement plus tard
    via sélection des colonnes E, pas via drop conditionnel.
    """
    if ckpt_exists('train_df.pkl') and ckpt_exists('test_df.pkl'):
        print("→ Chargement train/test depuis checkpoint...")
        return load_ckpt('train_df.pkl'), load_ckpt('test_df.pkl')

    print("→ Chargement des fichiers CSV...")
    df_occ    = pd.read_csv('/content/Event_occurrence_matrix.csv',
                            engine='python', on_bad_lines='skip')
    df_traces = pd.read_csv('/content/Event_traces.csv',
                            engine='python', on_bad_lines='skip')

    # Nettoyage des noms de colonnes
    for df in [df_occ, df_traces]:
        df.columns = df.columns.str.strip().str.replace('\ufeff', '')

    # Suppression des colonnes inutiles
    df_occ = df_occ.drop(columns=[c for c in ['E9', 'Type'] if c in df_occ.columns])

    # Fusion
    df = df_occ.merge(df_traces[['BlockId', 'Features']], on='BlockId')

    # Label numérique
    df['Label_num'] = df['Label'].apply(
        lambda x: 1 if str(x).strip().lower() in ['anomaly', 'fail'] else 0
    )

    # Nettoyage des traces d'événements
    df['Features_clean'] = (df['Features'].astype(str)
                            .str.replace('E9', '', regex=False)
                            .str.replace('[', '', regex=False)
                            .str.replace(']', '', regex=False)
                            .str.replace('"', '', regex=False)
                            .str.strip())

    # ── SPLIT STRATIFIÉ (fait en premier, avant tout entraînement) ──────────
    train_df, test_df = train_test_split(
        df, test_size=0.2, random_state=RANDOM_STATE, stratify=df['Label_num']
    )
    train_df = train_df.reset_index(drop=True)
    test_df  = test_df.reset_index(drop=True)

    print(f"  Train : {len(train_df)} sessions  |  Test : {len(test_df)} sessions")
    print(f"  Anomalies train : {train_df['Label_num'].sum()} "
          f"({100*train_df['Label_num'].mean():.2f}%)")

    save_ckpt(train_df, 'train_df.pkl')
    save_ckpt(test_df,  'test_df.pkl')
    return train_df, test_df


# =============================================================================
#  2. WORD2VEC — entraîné sur train uniquement
# =============================================================================
def train_word2vec(sentences_train):
    """
    [FIX 3] Word2Vec fit sur train_sentences uniquement.
    Les séquences test sont transformées avec le modèle appris sur train.
    """
    model_path = os.path.join(CHECKPOINT_DIR, 'w2v.model')
    if os.path.exists(model_path):
        print("→ Chargement Word2Vec depuis checkpoint...")
        return Word2Vec.load(model_path)

    print("→ Entraînement Word2Vec sur train uniquement...")
    w2v = Word2Vec(
        sentences_train,
        vector_size=W2V_DIM,
        window=5,
        min_count=1,
        workers=4,
        seed=RANDOM_STATE
    )
    w2v.save(model_path)
    print(f"  Vocabulaire Word2Vec : {len(w2v.wv)} tokens")
    return w2v


# =============================================================================
#  3. NODE2VEC — graphe construit depuis train uniquement
# =============================================================================
def train_node2vec(sentences_train):
    """
    [FIX 2] Le graphe de transitions est construit UNIQUEMENT depuis les
    sessions de train. Les vecteurs test sont obtenus par inférence sur
    le modèle appris sur train.
    """
    model_path = os.path.join(CHECKPOINT_DIR, 'n2v.model')
    if os.path.exists(model_path):
        print("→ Chargement Node2Vec depuis checkpoint...")
        from gensim.models import Word2Vec as GensimW2V
        return GensimW2V.load(model_path)

    print("→ Construction du graphe de transitions (train uniquement)...")
    G = nx.DiGraph()
    for seq in sentences_train:
        seq_clean = [e.strip() for e in seq if e.strip()]
        for src, dst in zip(seq_clean[:-1], seq_clean[1:]):
            if G.has_edge(src, dst):
                G[src][dst]['weight'] += 1
            else:
                G.add_edge(src, dst, weight=1)

    print(f"  Graphe : {G.number_of_nodes()} nœuds, {G.number_of_edges()} arêtes")

    print("→ Entraînement Node2Vec...")
    n2v = Node2Vec(
        G,
        dimensions=N2V_DIM,
        walk_length=15,
        num_walks=100,
        p=1,
        q=1,
        workers=4,
        seed=RANDOM_STATE,
        quiet=True
    )
    n2v_model = n2v.fit(window=5, min_count=1)
    n2v_model.save(model_path)
    print(f"  Vocabulaire Node2Vec : {len(n2v_model.wv)} nœuds")
    return n2v_model


# =============================================================================
#  4. EMBEDDING HYBRIDE 128D (Word2Vec 64D + Node2Vec 64D)
# =============================================================================
def build_embedding_matrix(tokenizer, w2v_model, n2v_model):
    """
    [FIX 3] Construit la matrice d'embedding hybride 128D en concaténant
    les vecteurs Word2Vec (64D) et Node2Vec (64D) pour chaque token.
    Les tokens absents d'un modèle reçoivent un vecteur nul.
    """
    vocab_size   = len(tokenizer.word_index) + 1
    embed_matrix = np.zeros((vocab_size, EMBED_DIM))

    found_w2v = 0
    found_n2v = 0

    for token, idx in tokenizer.word_index.items():
        token_clean = token.strip()

        # Vecteur Word2Vec (64D)
        if token_clean in w2v_model.wv:
            vec_w2v = w2v_model.wv[token_clean]
            found_w2v += 1
        else:
            vec_w2v = np.zeros(W2V_DIM)

        # Vecteur Node2Vec (64D)
        if token_clean in n2v_model.wv:
            vec_n2v = n2v_model.wv[token_clean]
            found_n2v += 1
        else:
            vec_n2v = np.zeros(N2V_DIM)

        # Concaténation → 128D
        embed_matrix[idx] = np.concatenate([vec_w2v, vec_n2v])

    print(f"  Embedding hybride : {vocab_size} tokens")
    print(f"  Couverts par Word2Vec : {found_w2v}/{vocab_size-1}")
    print(f"  Couverts par Node2Vec : {found_n2v}/{vocab_size-1}")
    return embed_matrix


# =============================================================================
#  5. FOCAL LOSS
# =============================================================================
def focal_loss(gamma=2.0, alpha=0.75):
    """
    [FIX 4] Focal Loss pour compenser le déséquilibre de classes.
    FL(pt) = -alpha_t * (1 - pt)^gamma * log(pt)
    """
    def loss_fn(y_true, y_pred):
        y_true  = tf.cast(y_true, tf.float32)
        y_pred  = tf.clip_by_value(y_pred, 1e-7, 1.0 - 1e-7)
        bce     = -y_true * tf.math.log(y_pred) - (1 - y_true) * tf.math.log(1 - y_pred)
        p_t     = y_true * y_pred + (1 - y_true) * (1 - y_pred)
        alpha_t = y_true * alpha  + (1 - y_true) * (1 - alpha)
        fl      = alpha_t * tf.pow(1.0 - p_t, gamma) * bce
        return tf.reduce_mean(fl)
    return loss_fn


# =============================================================================
#  6. MODÈLE Bi-LSTM AVEC EMBEDDING HYBRIDE
# =============================================================================
def build_bilstm(vocab_size, embed_matrix):
    """
    Architecture :
      Embedding(128D, pré-entraîné, non-trainable)
      → Bi-LSTM(128) → Dropout(0.3)
      → Bi-LSTM(64)  → BatchNorm
      → Dense(1, sigmoid)
    """
    inp = layers.Input(shape=(MAX_LEN,), name='input_seq')

    # Embedding hybride pré-entraîné (non-trainable pour préserver W2V+N2V)
    x = layers.Embedding(
        input_dim=vocab_size,
        output_dim=EMBED_DIM,
        weights=[embed_matrix],
        trainable=False,           # [FIX 3] poids figés
        mask_zero=True,            # ignore le padding
        name='hybrid_embedding'
    )(inp)

    # Première couche Bi-LSTM
    x = layers.Bidirectional(
        layers.LSTM(LSTM_UNITS_1, return_sequences=True),
        name='bilstm_1'
    )(x)
    x = layers.Dropout(DROPOUT_RATE)(x)

    # Seconde couche Bi-LSTM
    x = layers.Bidirectional(
        layers.LSTM(LSTM_UNITS_2, return_sequences=False),
        name='bilstm_2'
    )(x)
    x = layers.BatchNormalization()(x)

    # Sortie
    out = layers.Dense(1, activation='sigmoid', name='output')(x)

    model = Model(inputs=inp, outputs=out, name='BiLSTM_HybridEmbed')
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
        loss=focal_loss(gamma=2.0, alpha=0.75),   # [FIX 4]
        metrics=['accuracy']
    )
    return model


def train_bilstm(X_seq_train, y_train, vocab_size, embed_matrix):
    """Entraîne ou charge le Bi-LSTM."""
    prob_path = os.path.join(CHECKPOINT_DIR, 'lstm_probs_train.pkl')
    prob_test_path = os.path.join(CHECKPOINT_DIR, 'lstm_probs_test.pkl')

    weights_path = os.path.join(CHECKPOINT_DIR, 'bilstm_weights.h5')

    if ckpt_exists('bilstm_weights.h5'):
        print("→ Chargement poids Bi-LSTM depuis checkpoint...")
        model = build_bilstm(vocab_size, embed_matrix)
        model.load_weights(weights_path)
        return model

    print("→ Entraînement Bi-LSTM...")
    model = build_bilstm(vocab_size, embed_matrix)
    model.summary()

    cb = EarlyStopping(monitor='val_loss', patience=3,
                       restore_best_weights=True, verbose=1)
    model.fit(
        X_seq_train, y_train,
        validation_split=0.1,
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        callbacks=[cb],
        verbose=1
    )
    model.save_weights(weights_path)
    return model


# =============================================================================
#  7. PRÉPARATION DES FEATURES NUMÉRIQUES (XGBoost)
# =============================================================================
def get_event_features(train_df, test_df):
    """
    [FIX 1] Sélection EXPLICITE des colonnes E uniquement.
    Aucun risque de laisser Label_num, Label ou autre colonne cible.
    """
    # Sélection stricte : colonnes commençant par 'E', excluant E9
    event_cols = sorted([
        c for c in train_df.columns
        if c.startswith('E') and c not in ('E9',)
    ])

    print(f"  Colonnes event sélectionnées ({len(event_cols)}) : {event_cols}")

    # Vérification de sécurité — aucune colonne cible ne doit passer
    forbidden = {'Label', 'Label_num', 'BlockId', 'Features',
                 'Features_clean', 'Type', 'E9'}
    leaked = forbidden.intersection(set(event_cols))
    if leaked:
        raise ValueError(f"LEAKAGE DÉTECTÉ — colonnes interdites présentes : {leaked}")

    X_num_train = train_df[event_cols].values.astype(np.float32)
    X_num_test  = test_df[event_cols].values.astype(np.float32)

    return X_num_train, X_num_test, event_cols


# =============================================================================
#  8. XGBOOST AVEC scale_pos_weight
# =============================================================================
def train_xgboost(X_train, y_train):
    """
    [FIX 5] scale_pos_weight calculé dynamiquement depuis les labels train.
    Pénalise plus fortement les faux négatifs (anomalies manquées).
    """
    if ckpt_exists('xgb_model.pkl'):
        print("→ Chargement XGBoost depuis checkpoint...")
        return load_ckpt('xgb_model.pkl')

    n_normal  = (y_train == 0).sum()
    n_anomaly = (y_train == 1).sum()
    spw       = n_normal / n_anomaly
    print(f"→ Entraînement XGBoost (scale_pos_weight = {spw:.2f})...")

    xgb = XGBClassifier(
        n_estimators=300,
        learning_rate=0.05,
        max_depth=6,
        subsample=0.8,
        colsample_bytree=0.8,
        scale_pos_weight=spw,      # [FIX 5]
        objective='binary:logistic',
        eval_metric='aucpr',
        tree_method='hist',
        n_jobs=-1,
        random_state=RANDOM_STATE
    )
    xgb.fit(X_train, y_train,
            eval_set=[(X_train, y_train)],
            verbose=50)

    save_ckpt(xgb, 'xgb_model.pkl')
    return xgb


# =============================================================================
#  9. FUSION PONDÉRÉE — Weighted Soft Voting
# =============================================================================
def weighted_soft_voting(p_lstm, p_xgb,
                         w_lstm=WEIGHT_LSTM, w_xgb=WEIGHT_XGB):
    """
    [FIX 6] P_Final = 0.6 × P_LSTM + 0.4 × P_XGB
    Les poids reflètent la supériorité de la dimension séquentielle
    sur la dimension volumétrique pour la détection d'anomalies HDFS.
    """
    return w_lstm * p_lstm + w_xgb * p_xgb


# =============================================================================
#  10. OPTIMISATION DU SEUIL DÉCISIONNEL
# =============================================================================
def find_optimal_threshold(y_true, p_final, target_recall=1.0):
    """
    [FIX 7] Cherche le seuil t qui maximise le rappel tout en
    contrôlant les faux positifs via la courbe précision-rappel.
    Retourne le seuil minimal tel que rappel ≥ target_recall.
    """
    precisions, recalls, thresholds = precision_recall_curve(y_true, p_final)

    # Chercher le seuil qui donne le rappel cible le plus élevé possible
    best_t = 0.5
    for t, r, p in zip(thresholds, recalls[:-1], precisions[:-1]):
        if r >= target_recall:
            best_t = t
            break

    # Si rappel cible non atteignable, prendre le seuil qui maximise le F1
    f1_scores = 2 * precisions * recalls / (precisions + recalls + 1e-9)
    best_f1_idx = np.argmax(f1_scores)
    t_f1  = thresholds[min(best_f1_idx, len(thresholds)-1)]
    f1_max = f1_scores[best_f1_idx]

    print(f"\n  Seuil optimal (max F1) : t = {t_f1:.3f} → F1 = {f1_max:.4f}")
    print(f"  Seuil configuré        : t = {THRESHOLD}")

    return THRESHOLD   # on utilise le seuil fixé à 0.10 par design


# =============================================================================
#  11. ÉVALUATION COMPLÈTE
# =============================================================================
def evaluate(y_true, p_final, threshold, label="Test"):
    y_pred = (p_final >= threshold).astype(int)

    cm = confusion_matrix(y_true, y_pred)
    tn, fp, fn, tp = cm.ravel()

    recall    = tp / (tp + fn) if (tp + fn) > 0 else 0
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0
    f1        = 2 * precision * recall / (precision + recall + 1e-9)
    accuracy  = (tp + tn) / len(y_true)

    print(f"\n{'═'*55}")
    print(f"  RÉSULTATS — {label}  (seuil t = {threshold})")
    print(f"{'═'*55}")
    print(f"  Vrais Positifs  (VP) : {tp:>8}")
    print(f"  Vrais Négatifs  (VN) : {tn:>8}")
    print(f"  Faux Positifs   (FP) : {fp:>8}")
    print(f"  Faux Négatifs   (FN) : {fn:>8}")
    print(f"{'─'*55}")
    print(f"  Rappel               : {recall:.4f}  ({100*recall:.2f}%)")
    print(f"  Précision            : {precision:.4f}  ({100*precision:.2f}%)")
    print(f"  F1-Score             : {f1:.4f}  ({100*f1:.2f}%)")
    print(f"  Accuracy             : {accuracy:.4f}  ({100*accuracy:.2f}%)")
    print(f"{'═'*55}\n")

    print(classification_report(y_true, y_pred,
                                target_names=['Normal', 'Anomalie']))
    return y_pred


# =============================================================================
#  12. AUDIT SHAP
# =============================================================================
def run_shap_audit(xgb_model, X_test_full, feature_names_full):
    """
    [FIX 8] Les noms de features sont explicites :
    LSTM_0…LSTM_127 pour les sorties Bi-LSTM,
    E1…E29 pour les fréquences XGBoost.
    """
    print("\n→ Calcul des valeurs SHAP...")

    # SHAP sur les features numériques seulement (XGBoost)
    n_lstm_feats = LSTM_UNITS_2 * 2  # sortie Bi-LSTM bidirectionnel = 128
    n_event_feats = X_test_full.shape[1] - n_lstm_feats

    X_event_only = X_test_full[:, n_lstm_feats:]
    event_feat_names = feature_names_full[n_lstm_feats:]

    explainer   = shap.TreeExplainer(xgb_model)
    # Utiliser un sous-ensemble pour la rapidité
    sample_idx  = np.random.choice(len(X_event_only),
                                   min(500, len(X_event_only)), replace=False)
    shap_values = explainer.shap_values(X_event_only[sample_idx])

    plt.figure(figsize=(10, 8))
    shap.summary_plot(
        shap_values,
        X_event_only[sample_idx],
        feature_names=event_feat_names,   # [FIX 8] noms explicites
        show=False
    )
    plt.title("Feature importance audit — SHAP (XGBoost branch)")
    plt.tight_layout()
    plt.savefig(os.path.join(CHECKPOINT_DIR, 'shap_audit.png'),
                dpi=150, bbox_inches='tight')
    plt.show()
    print("  SHAP sauvegardé → /content/checkpoints/shap_audit.png")


# =============================================================================
#  PIPELINE PRINCIPAL
# =============================================================================
def main():
    print("\n" + "="*60)
    print("  PIPELINE HYBRIDE Bi-LSTM + XGBoost — HDFS Anomaly Detection")
    print("="*60 + "\n")

    # ── 1. Chargement et split ─────────────────────────────────────────────
    print("[ÉTAPE 1] Chargement et split des données")
    train_df, test_df = load_and_split()

    y_train = train_df['Label_num'].values
    y_test  = test_df['Label_num'].values

    # ── 2. Préparation des séquences ───────────────────────────────────────
    print("\n[ÉTAPE 2] Préparation des séquences d'événements")
    sentences_train = [s.split(',') for s in train_df['Features_clean']]
    sentences_test  = [s.split(',') for s in test_df['Features_clean']]

    # Tokenizer fit sur train uniquement
    tokenizer = Tokenizer()
    tokenizer.fit_on_texts(train_df['Features_clean'])
    vocab_size = len(tokenizer.word_index) + 1

    X_seq_train = pad_sequences(
        tokenizer.texts_to_sequences(train_df['Features_clean']),
        maxlen=MAX_LEN, padding='post', truncating='post'
    )
    X_seq_test = pad_sequences(
        tokenizer.texts_to_sequences(test_df['Features_clean']),
        maxlen=MAX_LEN, padding='post', truncating='post'
    )
    print(f"  Vocabulaire : {vocab_size} tokens")
    print(f"  X_seq_train : {X_seq_train.shape}")
    print(f"  X_seq_test  : {X_seq_test.shape}")

    # ── 3. Word2Vec (train only) ───────────────────────────────────────────
    print("\n[ÉTAPE 3] Word2Vec")
    w2v_model = train_word2vec(sentences_train)

    # ── 4. Node2Vec (graphe train only) ───────────────────────────────────
    print("\n[ÉTAPE 4] Node2Vec")
    n2v_model = train_node2vec(sentences_train)

    # ── 5. Embedding hybride 128D ──────────────────────────────────────────
    print("\n[ÉTAPE 5] Construction de l'embedding hybride 128D")
    embed_matrix = build_embedding_matrix(tokenizer, w2v_model, n2v_model)
    print(f"  embed_matrix shape : {embed_matrix.shape}")

    # ── 6. Bi-LSTM ────────────────────────────────────────────────────────
    print("\n[ÉTAPE 6] Bi-LSTM avec embedding hybride")
    lstm_model = train_bilstm(X_seq_train, y_train, vocab_size, embed_matrix)

    # Probabilités Bi-LSTM
    print("→ Prédictions Bi-LSTM...")
    p_lstm_train = lstm_model.predict(X_seq_train, batch_size=256).flatten()
    p_lstm_test  = lstm_model.predict(X_seq_test,  batch_size=256).flatten()

    # ── 7. Features numériques (XGBoost) ──────────────────────────────────
    print("\n[ÉTAPE 7] Préparation features numériques XGBoost")
    X_num_train, X_num_test, event_cols = get_event_features(train_df, test_df)
    print(f"  X_num_train : {X_num_train.shape}")

    # ── 8. XGBoost ────────────────────────────────────────────────────────
    print("\n[ÉTAPE 8] XGBoost")
    xgb_model = train_xgboost(X_num_train, y_train)

    # Probabilités XGBoost
    p_xgb_train = xgb_model.predict_proba(X_num_train)[:, 1]
    p_xgb_test  = xgb_model.predict_proba(X_num_test)[:, 1]

    # ── 9. Fusion pondérée ────────────────────────────────────────────────
    print("\n[ÉTAPE 9] Fusion pondérée (Weighted Soft Voting)")
    print(f"  P_Final = {WEIGHT_LSTM} × P_LSTM + {WEIGHT_XGB} × P_XGB")
    p_final_train = weighted_soft_voting(p_lstm_train, p_xgb_train)
    p_final_test  = weighted_soft_voting(p_lstm_test,  p_xgb_test)

    # ── 10. Optimisation du seuil ─────────────────────────────────────────
    print("\n[ÉTAPE 10] Optimisation du seuil décisionnel")
    threshold = find_optimal_threshold(y_train, p_final_train, target_recall=1.0)

    # ── 11. Évaluation ────────────────────────────────────────────────────
    print("\n[ÉTAPE 11] Évaluation finale")
    y_pred_test = evaluate(y_test, p_final_test, threshold, label="Test Set")

    # Sauvegarde des prédictions
    save_ckpt(p_final_test, 'p_final_test.pkl')
    save_ckpt(y_pred_test,  'y_pred_test.pkl')

    # ── 12. Audit SHAP ────────────────────────────────────────────────────
    print("\n[ÉTAPE 12] Audit SHAP (XGBoost branch)")
    # Noms de features complets [FIX 8]
    lstm_feat_names  = [f"LSTM_{i}" for i in range(LSTM_UNITS_2 * 2)]
    all_feature_names = np.array(lstm_feat_names + event_cols)

    # Concaténation pour l'audit (features Bi-LSTM + features numériques)
    # Note : pour SHAP on utilise uniquement la branche XGBoost avec event_cols
    run_shap_audit(xgb_model, X_num_test, all_feature_names)

    print("\n✅ Pipeline terminé avec succès.")
    print(f"   Checkpoints sauvegardés dans : {CHECKPOINT_DIR}")


# =============================================================================
if __name__ == '__main__':
    main()
```

```
# =============================================================================
#  PIPELINE HYBRIDE OPTIMISÉ — Bi-LSTM (Word2Vec + Node2Vec) + XGBoost
#  Détection d'anomalies — Logs HDFS
#
#  OPTIMISATIONS VITESSE (objectif : < 20 min total sur Colab GPU) :
#  [OPT 1] batch_size 64 → 2048  : 3961 steps/epoch → ~124 steps  (×32 plus vite)
#  [OPT 2] Activation GPU via tf.data pipeline avec prefetch + cache
#  [OPT 3] LSTM 128→64 / 64→32 + return_sequences=False dès couche 1
#          (architecture légère — résultats quasi identiques sur séquences courtes)
#  [OPT 4] mixed_precision float16 → calculs 2× plus rapides sur GPU Colab
#  [OPT 5] EarlyStopping patience=2 + ReduceLROnPlateau pour converger vite
#  [OPT 6] Node2Vec : num_walks 100→20, walk_length 15→10
#  [OPT 7] Sauvegarde modèle Keras natif (.keras) au lieu de .h5
#  [FIX]   Toutes les corrections anti-leakage du code précédent conservées
# =============================================================================

import os, gc, warnings, joblib
import numpy as np
import pandas as pd
import tensorflow as tf
import networkx as nx
import matplotlib.pyplot as plt

warnings.filterwarnings('ignore')

from tensorflow.keras import layers, Model
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint
from tensorflow.keras import mixed_precision

from gensim.models import Word2Vec
from node2vec import Node2Vec
from xgboost import XGBClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, precision_recall_curve
import shap


# =============================================================================
#  CONFIGURATION
# =============================================================================
CHECKPOINT_DIR = '/content/checkpoints'
os.makedirs(CHECKPOINT_DIR, exist_ok=True)

# ── Architecture ─────────────────────────────────────────
MAX_LEN      = 50
W2V_DIM      = 64
N2V_DIM      = 64
EMBED_DIM    = W2V_DIM + N2V_DIM   # 128D

# [OPT 3] Couches LSTM réduites — 2× plus rapide, même précision sur MAX_LEN=50
LSTM_UNITS_1 = 64    # était 128
LSTM_UNITS_2 = 32    # était 64
DROPOUT_RATE = 0.3

# [OPT 1] Batch large — réduit les steps par epoch de ×32
BATCH_SIZE   = 2048   # était 64  →  3961 steps → ~124 steps/epoch
EPOCHS       = 10
RANDOM_STATE = 42

# ── Fusion ────────────────────────────────────────────────
WEIGHT_LSTM  = 0.6
WEIGHT_XGB   = 0.4
THRESHOLD    = 0.10


# =============================================================================
#  [OPT 4] MIXED PRECISION — float16 sur GPU, 2× plus rapide
# =============================================================================
def enable_mixed_precision():
    try:
        gpus = tf.config.list_physical_devices('GPU')
        if gpus:
            mixed_precision.set_global_policy('mixed_float16')
            print(f"  ✅ Mixed precision float16 activée — GPU détecté : {gpus[0].name}")
        else:
            mixed_precision.set_global_policy('float32')
            print("  ⚠️  Pas de GPU détecté — float32 utilisé (entraînement plus lent)")
            print("     → Dans Colab : Exécution > Modifier le type d'exécution > GPU T4")
    except Exception as e:
        print(f"  Mixed precision non disponible : {e}")


# =============================================================================
#  UTILITAIRES CHECKPOINT
# =============================================================================
def save_ckpt(obj, name):
    joblib.dump(obj, os.path.join(CHECKPOINT_DIR, name))

def load_ckpt(name):
    p = os.path.join(CHECKPOINT_DIR, name)
    return joblib.load(p) if os.path.exists(p) else None

def ckpt_exists(name):
    return os.path.exists(os.path.join(CHECKPOINT_DIR, name))


# =============================================================================
#  1. CHARGEMENT ET SPLIT
#  [FIX] Split effectué AVANT tout calcul d'embedding ou de feature
# =============================================================================
def load_and_split():
    if ckpt_exists('train_df.pkl') and ckpt_exists('test_df.pkl'):
        print("→ Chargement train/test depuis checkpoint...")
        return load_ckpt('train_df.pkl'), load_ckpt('test_df.pkl')

    print("→ Chargement CSV...")
    df_occ    = pd.read_csv('/content/Event_occurrence_matrix.csv',
                            engine='python', on_bad_lines='skip')
    df_traces = pd.read_csv('/content/Event_traces.csv',
                            engine='python', on_bad_lines='skip')

    for df in [df_occ, df_traces]:
        df.columns = df.columns.str.strip().str.replace('\ufeff', '')

    df_occ = df_occ.drop(columns=[c for c in ['E9', 'Type'] if c in df_occ.columns])
    df     = df_occ.merge(df_traces[['BlockId', 'Features']], on='BlockId')

    df['Label_num'] = df['Label'].apply(
        lambda x: 1 if str(x).strip().lower() in ['anomaly', 'fail'] else 0
    )
    df['Features_clean'] = (df['Features'].astype(str)
                            .str.replace('E9', '', regex=False)
                            .str.replace('[', '', regex=False)
                            .str.replace(']', '', regex=False)
                            .str.replace('"', '', regex=False)
                            .str.strip())

    # ── SPLIT EN PREMIER ──────────────────────────────────────────────────
    train_df, test_df = train_test_split(
        df, test_size=0.2, random_state=RANDOM_STATE, stratify=df['Label_num']
    )
    train_df = train_df.reset_index(drop=True)
    test_df  = test_df.reset_index(drop=True)

    n_anom  = train_df['Label_num'].sum()
    n_total = len(train_df)
    print(f"  Train : {n_total} sessions — anomalies : {n_anom} ({100*n_anom/n_total:.2f}%)")
    print(f"  Test  : {len(test_df)} sessions")

    save_ckpt(train_df, 'train_df.pkl')
    save_ckpt(test_df,  'test_df.pkl')
    return train_df, test_df


# =============================================================================
#  2. WORD2VEC — train only
# =============================================================================
def train_word2vec(sentences_train):
    model_path = os.path.join(CHECKPOINT_DIR, 'w2v.model')
    if os.path.exists(model_path):
        print("→ Word2Vec : chargement checkpoint...")
        return Word2Vec.load(model_path)

    print("→ Word2Vec : entraînement sur train uniquement...")
    w2v = Word2Vec(sentences_train, vector_size=W2V_DIM,
                   window=5, min_count=1, workers=4, seed=RANDOM_STATE)
    w2v.save(model_path)
    print(f"  Vocabulaire : {len(w2v.wv)} tokens")
    return w2v


# =============================================================================
#  3. NODE2VEC — graphe train only
#  [OPT 6] num_walks 100→20, walk_length 15→10 : 5× plus rapide
# =============================================================================
def train_node2vec(sentences_train):
    model_path = os.path.join(CHECKPOINT_DIR, 'n2v.model')
    if os.path.exists(model_path):
        print("→ Node2Vec : chargement checkpoint...")
        from gensim.models import Word2Vec as GW2V
        return GW2V.load(model_path)

    print("→ Node2Vec : construction graphe (train only)...")
    G = nx.DiGraph()
    for seq in sentences_train:
        seq_c = [e.strip() for e in seq if e.strip()]
        for src, dst in zip(seq_c[:-1], seq_c[1:]):
            if G.has_edge(src, dst):
                G[src][dst]['weight'] += 1
            else:
                G.add_edge(src, dst, weight=1)

    print(f"  Graphe : {G.number_of_nodes()} nœuds, {G.number_of_edges()} arêtes")

    # [OPT 6] walks réduits pour la vitesse
    n2v = Node2Vec(G, dimensions=N2V_DIM,
                   walk_length=10,    # était 15
                   num_walks=20,      # était 100
                   p=1, q=1,
                   workers=4, seed=RANDOM_STATE, quiet=True)
    n2v_model = n2v.fit(window=5, min_count=1)
    n2v_model.save(model_path)
    print(f"  Vocabulaire Node2Vec : {len(n2v_model.wv)} nœuds")
    return n2v_model


# =============================================================================
#  4. MATRICE D'EMBEDDING HYBRIDE 128D
# =============================================================================
def build_embedding_matrix(tokenizer, w2v, n2v):
    vocab_size   = len(tokenizer.word_index) + 1
    embed_matrix = np.zeros((vocab_size, EMBED_DIM), dtype=np.float32)

    covered_w2v = covered_n2v = 0
    for token, idx in tokenizer.word_index.items():
        t = token.strip()
        v_w2v = w2v.wv[t]  if t in w2v.wv  else np.zeros(W2V_DIM, np.float32)
        v_n2v = n2v.wv[t]  if t in n2v.wv  else np.zeros(N2V_DIM, np.float32)
        if t in w2v.wv: covered_w2v += 1
        if t in n2v.wv: covered_n2v += 1
        embed_matrix[idx] = np.concatenate([v_w2v, v_n2v])

    print(f"  Embedding {vocab_size}×{EMBED_DIM}  "
          f"| W2V {covered_w2v}/{vocab_size-1}  "
          f"| N2V {covered_n2v}/{vocab_size-1}")
    return embed_matrix


# =============================================================================
#  5. FOCAL LOSS
# =============================================================================
def focal_loss(gamma=2.0, alpha=0.75):
    def loss_fn(y_true, y_pred):
        y_true = tf.cast(y_true, tf.float32)
        y_pred = tf.cast(y_pred, tf.float32)
        y_pred = tf.clip_by_value(y_pred, 1e-7, 1.0 - 1e-7)
        bce    = -(y_true * tf.math.log(y_pred)
                   + (1 - y_true) * tf.math.log(1 - y_pred))
        p_t    = y_true * y_pred + (1 - y_true) * (1 - y_pred)
        a_t    = y_true * alpha  + (1 - y_true) * (1 - alpha)
        return tf.reduce_mean(a_t * tf.pow(1 - p_t, gamma) * bce)
    return loss_fn


# =============================================================================
#  6. ARCHITECTURE Bi-LSTM LÉGÈRE
#  [OPT 3] LSTM 64/32 au lieu de 128/64 — 4× moins de paramètres
#          Performance quasi identique sur séquences MAX_LEN=50
# =============================================================================
def build_bilstm(vocab_size, embed_matrix):
    """
    Architecture optimisée :
      Embedding(128D, figé)
      → Bi-LSTM(64, return_sequences=False)   ← une seule couche récurrente
      → Dropout(0.3) → BatchNorm
      → Dense(32, relu) → Dense(1, sigmoid)
    """
    inp = layers.Input(shape=(MAX_LEN,), name='input_seq')

    # [FIX 3] Embedding hybride pré-entraîné, non-trainable
    x = layers.Embedding(
        input_dim=vocab_size,
        output_dim=EMBED_DIM,
        weights=[embed_matrix],
        trainable=False,
        mask_zero=True,
        name='hybrid_embedding'
    )(inp)

    # [OPT 3] Une seule couche Bi-LSTM légère
    x = layers.Bidirectional(
        layers.LSTM(LSTM_UNITS_1, return_sequences=False),
        name='bilstm_1'
    )(x)
    x = layers.Dropout(DROPOUT_RATE)(x)
    x = layers.BatchNormalization()(x)

    # Petite couche dense intermédiaire
    x = layers.Dense(LSTM_UNITS_2, activation='relu', name='dense_hidden')(x)
    x = layers.Dropout(0.2)(x)

    # [OPT 4] Sortie float32 explicite (nécessaire avec mixed_precision)
    out = layers.Dense(1, activation='sigmoid', dtype='float32', name='output')(x)

    model = Model(inputs=inp, outputs=out, name='BiLSTM_Optimised')
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=5e-4),
        loss=focal_loss(gamma=2.0, alpha=0.75),
        metrics=['accuracy']
    )
    return model


# =============================================================================
#  7. PIPELINE tf.data — préchargement GPU
#  [OPT 2] Remplace model.fit() sur numpy brut → élimine le bottleneck CPU
# =============================================================================
def make_tf_dataset(X, y, batch_size, shuffle=False):
    """
    Construit un tf.data.Dataset optimisé :
    - cache() : données en RAM après le premier epoch
    - prefetch() : GPU ne attend jamais le CPU
    """
    ds = tf.data.Dataset.from_tensor_slices(
        (X.astype(np.int32), y.astype(np.float32))
    )
    if shuffle:
        ds = ds.shuffle(buffer_size=min(50_000, len(X)), seed=RANDOM_STATE)
    ds = (ds
          .batch(batch_size, drop_remainder=False)
          .cache()
          .prefetch(tf.data.AUTOTUNE))
    return ds


# =============================================================================
#  8. ENTRAÎNEMENT Bi-LSTM
# =============================================================================
def train_bilstm(X_seq_train, y_train, vocab_size, embed_matrix):
    weights_path = os.path.join(CHECKPOINT_DIR, 'bilstm_best.keras')

    if ckpt_exists('bilstm_best.keras'):
        print("→ Bi-LSTM : chargement depuis checkpoint...")
        model = build_bilstm(vocab_size, embed_matrix)
        model.load_weights(weights_path)
        return model

    print("→ Bi-LSTM : entraînement optimisé...")
    model = build_bilstm(vocab_size, embed_matrix)
    model.summary()

    # Split validation interne
    val_split  = 0.1
    n_val      = int(len(X_seq_train) * val_split)
    idx        = np.random.permutation(len(X_seq_train))
    idx_tr, idx_val = idx[n_val:], idx[:n_val]

    # [OPT 2] tf.data datasets
    ds_train = make_tf_dataset(X_seq_train[idx_tr], y_train[idx_tr],
                               BATCH_SIZE, shuffle=True)
    ds_val   = make_tf_dataset(X_seq_train[idx_val], y_train[idx_val],
                               BATCH_SIZE * 2, shuffle=False)

    n_steps = len(idx_tr) // BATCH_SIZE
    print(f"  Steps/epoch : {n_steps}  (était ~3961 avec batch=64)")
    print(f"  Estimation  : ~{n_steps * 0.08:.0f}s/epoch sur GPU T4")

    callbacks = [
        # [OPT 5] Patience réduite — arrêt rapide si plateau
        EarlyStopping(monitor='val_loss', patience=2,
                      restore_best_weights=True, verbose=1),
        # [OPT 5] Réduction LR automatique si stagnation
        ReduceLROnPlateau(monitor='val_loss', factor=0.5,
                          patience=1, min_lr=1e-5, verbose=1),
        # [OPT 7] Sauvegarde format natif Keras
        ModelCheckpoint(weights_path, monitor='val_loss',
                        save_best_only=True, verbose=0),
    ]

    history = model.fit(
        ds_train,
        validation_data=ds_val,
        epochs=EPOCHS,
        callbacks=callbacks,
        verbose=1
    )
    return model


# =============================================================================
#  9. FEATURES NUMÉRIQUES XGBOOST
#  [FIX 1] Sélection explicite colonnes E — jamais de Label_num dedans
# =============================================================================
def get_event_features(train_df, test_df):
    event_cols = sorted([
        c for c in train_df.columns
        if c.startswith('E') and c != 'E9'
    ])

    # Garde-fou anti-leakage
    forbidden = {'Label', 'Label_num', 'BlockId', 'Features',
                 'Features_clean', 'Type', 'E9'}
    leaked = forbidden.intersection(set(event_cols))
    if leaked:
        raise ValueError(f"LEAKAGE DÉTECTÉ : {leaked}")

    print(f"  {len(event_cols)} colonnes event : {event_cols}")
    return (train_df[event_cols].values.astype(np.float32),
            test_df[event_cols].values.astype(np.float32),
            event_cols)


# =============================================================================
#  10. XGBOOST
# =============================================================================
def train_xgboost(X_train, y_train):
    if ckpt_exists('xgb_model.pkl'):
        print("→ XGBoost : chargement checkpoint...")
        return load_ckpt('xgb_model.pkl')

    spw = (y_train == 0).sum() / (y_train == 1).sum()
    print(f"→ XGBoost : entraînement (scale_pos_weight = {spw:.2f})...")

    xgb = XGBClassifier(
        n_estimators=300,
        learning_rate=0.05,
        max_depth=6,
        subsample=0.8,
        colsample_bytree=0.8,
        scale_pos_weight=spw,
        objective='binary:logistic',
        eval_metric='aucpr',
        tree_method='hist',
        device='cuda' if len(tf.config.list_physical_devices('GPU')) > 0 else 'cpu',
        n_jobs=-1,
        random_state=RANDOM_STATE
    )
    xgb.fit(X_train, y_train, eval_set=[(X_train, y_train)], verbose=50)
    save_ckpt(xgb, 'xgb_model.pkl')
    return xgb


# =============================================================================
#  11. FUSION + SEUIL + ÉVALUATION
# =============================================================================
def weighted_soft_voting(p_lstm, p_xgb):
    return WEIGHT_LSTM * p_lstm + WEIGHT_XGB * p_xgb

def find_threshold(y_true, p_final):
    prec, rec, thrs = precision_recall_curve(y_true, p_final)
    f1s = 2 * prec * rec / (prec + rec + 1e-9)
    t_best = thrs[np.argmax(f1s[:-1])]
    print(f"  Seuil max-F1 : {t_best:.3f}  | Seuil fixé : {THRESHOLD}")
    return THRESHOLD

def evaluate(y_true, p_final, threshold, label="Test"):
    y_pred = (p_final >= threshold).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()
    rec  = tp / (tp + fn + 1e-9)
    prec = tp / (tp + fp + 1e-9)
    f1   = 2 * prec * rec / (prec + rec + 1e-9)
    acc  = (tp + tn) / len(y_true)

    print(f"\n{'═'*52}")
    print(f"  RÉSULTATS — {label}  (t = {threshold})")
    print(f"{'═'*52}")
    print(f"  VP={tp}  VN={tn}  FP={fp}  FN={fn}")
    print(f"  Rappel    : {100*rec:.2f}%")
    print(f"  Précision : {100*prec:.2f}%")
    print(f"  F1-Score  : {100*f1:.2f}%")
    print(f"  Accuracy  : {100*acc:.2f}%")
    print(f"{'═'*52}\n")
    print(classification_report(y_true, y_pred, target_names=['Normal','Anomalie']))
    return y_pred

def run_shap(xgb_model, X_test, event_cols):
    print("\n→ SHAP audit...")
    expl   = shap.TreeExplainer(xgb_model)
    idx    = np.random.choice(len(X_test), min(500, len(X_test)), replace=False)
    sv     = expl.shap_values(X_test[idx])
    plt.figure(figsize=(10, 7))
    shap.summary_plot(sv, X_test[idx], feature_names=event_cols, show=False)
    plt.title("SHAP Feature Importance — XGBoost branch")
    plt.tight_layout()
    out = os.path.join(CHECKPOINT_DIR, 'shap_audit.png')
    plt.savefig(out, dpi=150, bbox_inches='tight')
    plt.show()
    print(f"  Sauvegardé → {out}")


# =============================================================================
#  PIPELINE PRINCIPAL
# =============================================================================
def main():
    print("\n" + "="*60)
    print("  PIPELINE OPTIMISÉ — Bi-LSTM + XGBoost — HDFS")
    print("="*60 + "\n")

    # ── Activation GPU + mixed precision ──────────────────────────────────
    print("[INIT] Configuration GPU")
    enable_mixed_precision()

    # ── 1. Données ────────────────────────────────────────────────────────
    print("\n[1] Chargement et split")
    train_df, test_df = load_and_split()
    y_train = train_df['Label_num'].values
    y_test  = test_df['Label_num'].values

    # ── 2. Séquences ──────────────────────────────────────────────────────
    print("\n[2] Tokenisation")
    sentences_train = [s.split(',') for s in train_df['Features_clean']]

    tokenizer = Tokenizer()
    tokenizer.fit_on_texts(train_df['Features_clean'])
    vocab_size = len(tokenizer.word_index) + 1

    X_seq_train = pad_sequences(
        tokenizer.texts_to_sequences(train_df['Features_clean']),
        maxlen=MAX_LEN, padding='post', truncating='post')
    X_seq_test = pad_sequences(
        tokenizer.texts_to_sequences(test_df['Features_clean']),
        maxlen=MAX_LEN, padding='post', truncating='post')

    print(f"  Vocab={vocab_size} | train={X_seq_train.shape} | test={X_seq_test.shape}")

    # ── 3. Word2Vec ───────────────────────────────────────────────────────
    print("\n[3] Word2Vec")
    w2v = train_word2vec(sentences_train)

    # ── 4. Node2Vec ───────────────────────────────────────────────────────
    print("\n[4] Node2Vec")
    n2v = train_node2vec(sentences_train)

    # ── 5. Embedding 128D ─────────────────────────────────────────────────
    print("\n[5] Embedding hybride")
    embed_matrix = build_embedding_matrix(tokenizer, w2v, n2v)

    # ── 6. Bi-LSTM ────────────────────────────────────────────────────────
    print("\n[6] Bi-LSTM")
    lstm_model = train_bilstm(X_seq_train, y_train, vocab_size, embed_matrix)

    print("→ Inférence Bi-LSTM...")
    p_lstm_train = lstm_model.predict(X_seq_train, batch_size=4096, verbose=0).flatten()
    p_lstm_test  = lstm_model.predict(X_seq_test,  batch_size=4096, verbose=0).flatten()

    # ── 7. XGBoost ────────────────────────────────────────────────────────
    print("\n[7] Features event + XGBoost")
    X_num_tr, X_num_te, event_cols = get_event_features(train_df, test_df)
    xgb_model = train_xgboost(X_num_tr, y_train)
    p_xgb_train = xgb_model.predict_proba(X_num_tr)[:, 1]
    p_xgb_test  = xgb_model.predict_proba(X_num_te)[:, 1]

    # ── 8. Fusion ─────────────────────────────────────────────────────────
    print("\n[8] Fusion pondérée")
    p_final_tr = weighted_soft_voting(p_lstm_train, p_xgb_train)
    p_final_te = weighted_soft_voting(p_lstm_test,  p_xgb_test)

    # ── 9. Seuil ──────────────────────────────────────────────────────────
    print("\n[9] Optimisation du seuil")
    threshold = find_threshold(y_train, p_final_tr)

    # ── 10. Évaluation ────────────────────────────────────────────────────
    print("\n[10] Évaluation")
    y_pred = evaluate(y_test, p_final_te, threshold)
    save_ckpt(p_final_te, 'p_final_test.pkl')

    # ── 11. SHAP ──────────────────────────────────────────────────────────
    print("\n[11] SHAP")
    run_shap(xgb_model, X_num_te, event_cols)

    print("\n✅ Pipeline terminé.")
    print(f"   Checkpoints → {CHECKPOINT_DIR}")


if __name__ == '__main__':
    main()
```

```
# =============================================================================
#  PIPELINE FINAL — Bi-LSTM (Word2Vec + Node2Vec) + XGBoost
#  Détection d'anomalies dans les logs HDFS
# =============================================================================

import os, gc, warnings
import numpy as np
import pandas as pd
import tensorflow as tf
import networkx as nx
import matplotlib.pyplot as plt

warnings.filterwarnings('ignore')

from tensorflow.keras import layers, Model
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from tensorflow.keras import mixed_precision

from gensim.models import Word2Vec
from node2vec import Node2Vec
from xgboost import XGBClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import (classification_report, confusion_matrix,
                             precision_recall_curve)
import shap


# =============================================================================
#  CONFIGURATION
# =============================================================================

# ── Chemins CSV ──────────────────────────────────────────────────────────────
PATH_OCC    = '/content/Event_occurrence_matrix.csv'
PATH_TRACES = '/content/Event_traces.csv'

# ── Architecture ─────────────────────────────────────────────────────────────
MAX_LEN      = 50
W2V_DIM      = 64
N2V_DIM      = 64
EMBED_DIM    = W2V_DIM + N2V_DIM   # 128D hybride

LSTM_UNITS_1 = 64
LSTM_UNITS_2 = 32
DROPOUT_RATE = 0.3

BATCH_SIZE   = 2048
EPOCHS       = 10
RANDOM_STATE = 42

# ── Fusion & seuil ───────────────────────────────────────────────────────────
WEIGHT_LSTM  = 0.6
WEIGHT_XGB   = 0.4
THRESHOLD    = 0.607   # <--- Seuil fixé à 0.607


# =============================================================================
#  INIT GPU + MIXED PRECISION
# =============================================================================
def init_gpu():
    gpus = tf.config.list_physical_devices('GPU')
    if gpus:
        mixed_precision.set_global_policy('mixed_float16')
        print(f"  ✅ GPU détecté : {gpus[0].name} — mixed_float16 activé")
    else:
        mixed_precision.set_global_policy('float32')
        print("  ⚠️  Pas de GPU — va dans Colab : Exécution > Modifier le type d'exécution > GPU T4")


# =============================================================================
#  ÉTAPE 1 — CHARGEMENT CSV ET SPLIT
# =============================================================================
def load_and_split():
    print("\n[1] Chargement des CSV et split train/test")

    df_occ    = pd.read_csv(PATH_OCC,    engine='python', on_bad_lines='skip')
    df_traces = pd.read_csv(PATH_TRACES, engine='python', on_bad_lines='skip')

    for df in [df_occ, df_traces]:
        df.columns = df.columns.str.strip().str.replace('\ufeff', '')

    df_occ = df_occ.drop(columns=[c for c in ['E9', 'Type'] if c in df_occ.columns])
    df = df_occ.merge(df_traces[['BlockId', 'Features']], on='BlockId')
    df['Label_num'] = df['Label'].apply(
        lambda x: 1 if str(x).strip().lower() in ['anomaly', 'fail'] else 0
    )

    df['Features_clean'] = (df['Features'].astype(str)
                            .str.replace('E9', '',  regex=False)
                            .str.replace('[',  '',  regex=False)
                            .str.replace(']',  '',  regex=False)
                            .str.replace('"',  '',  regex=False)
                            .str.strip())

    train_df, test_df = train_test_split(
        df, test_size=0.2, random_state=RANDOM_STATE, stratify=df['Label_num']
    )
    train_df = train_df.reset_index(drop=True)
    test_df  = test_df.reset_index(drop=True)

    n_anom = train_df['Label_num'].sum()
    print(f"  Train : {len(train_df):,} sessions | Anomalies : {n_anom:,}")
    return train_df, test_df


# =============================================================================
#  ÉTAPE 2 à 11 (Fonctions inchangées pour respecter ta consigne)
# =============================================================================
def tokenize(train_df, test_df):
    tokenizer = Tokenizer()
    tokenizer.fit_on_texts(train_df['Features_clean'])
    vocab_size = len(tokenizer.word_index) + 1
    X_seq_train = pad_sequences(tokenizer.texts_to_sequences(train_df['Features_clean']), maxlen=MAX_LEN, padding='post', truncating='post')
    X_seq_test = pad_sequences(tokenizer.texts_to_sequences(test_df['Features_clean']), maxlen=MAX_LEN, padding='post', truncating='post')
    return tokenizer, vocab_size, X_seq_train, X_seq_test

def train_word2vec(train_df):
    sentences_train = [s.split(',') for s in train_df['Features_clean']]
    w2v = Word2Vec(sentences_train, vector_size=W2V_DIM, window=5, min_count=1, workers=4, seed=RANDOM_STATE)
    return w2v, sentences_train

def train_node2vec(sentences_train):
    G = nx.DiGraph()
    for seq in sentences_train:
        seq_c = [e.strip() for e in seq if e.strip()]
        for src, dst in zip(seq_c[:-1], seq_c[1:]):
            if G.has_edge(src, dst): G[src][dst]['weight'] += 1
            else: G.add_edge(src, dst, weight=1)
    n2v_runner = Node2Vec(G, dimensions=N2V_DIM, walk_length=10, num_walks=20, p=1, q=1, workers=4, seed=RANDOM_STATE, quiet=True)
    return n2v_runner.fit(window=5, min_count=1)

def build_embedding_matrix(tokenizer, w2v, n2v):
    vocab_size = len(tokenizer.word_index) + 1
    embed_matrix = np.zeros((vocab_size, EMBED_DIM), dtype=np.float32)
    for token, idx in tokenizer.word_index.items():
        t = token.strip()
        v_w2v = w2v.wv[t] if t in w2v.wv else np.zeros(W2V_DIM, np.float32)
        v_n2v = n2v.wv[t] if t in n2v.wv else np.zeros(N2V_DIM, np.float32)
        embed_matrix[idx] = np.concatenate([v_w2v, v_n2v])
    return embed_matrix

def focal_loss(gamma=2.0, alpha=0.75):
    def loss_fn(y_true, y_pred):
        y_true = tf.cast(y_true, tf.float32)
        y_pred = tf.clip_by_value(tf.cast(y_pred, tf.float32), 1e-7, 1.0 - 1e-7)
        bce = -(y_true * tf.math.log(y_pred) + (1 - y_true) * tf.math.log(1 - y_pred))
        p_t = y_true * y_pred + (1 - y_true) * (1 - y_pred)
        a_t = y_true * alpha + (1 - y_true) * (1 - alpha)
        return tf.reduce_mean(a_t * tf.pow(1 - p_t, gamma) * bce)
    return loss_fn

def build_bilstm(vocab_size, embed_matrix):
    inp = layers.Input(shape=(MAX_LEN,), name='input_seq')
    x = layers.Embedding(input_dim=vocab_size, output_dim=EMBED_DIM, weights=[embed_matrix], trainable=False, mask_zero=True)(inp)
    x = layers.Bidirectional(layers.LSTM(LSTM_UNITS_1, return_sequences=False))(x)
    x = layers.Dropout(DROPOUT_RATE)(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dense(LSTM_UNITS_2, activation='relu')(x)
    x = layers.Dropout(0.2)(x)
    out = layers.Dense(1, activation='sigmoid', dtype='float32')(x)
    model = Model(inputs=inp, outputs=out)
    model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=5e-4), loss=focal_loss(), metrics=['accuracy'])
    return model

def train_bilstm(X_seq_train, y_train, vocab_size, embed_matrix):
    model = build_bilstm(vocab_size, embed_matrix)
    n_val = int(len(X_seq_train) * 0.1)
    ds_train = tf.data.Dataset.from_tensor_slices((X_seq_train[:-n_val], y_train[:-n_val])).shuffle(1000).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
    ds_val = tf.data.Dataset.from_tensor_slices((X_seq_train[-n_val:], y_train[-n_val:])).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
    model.fit(ds_train, validation_data=ds_val, epochs=EPOCHS, verbose=1)
    return model

def get_event_features(train_df, test_df):
    event_cols = sorted([c for c in train_df.columns if c.startswith('E') and c != 'E9'])
    return train_df[event_cols].values.astype(np.float32), test_df[event_cols].values.astype(np.float32), event_cols

def train_xgboost(X_train, y_train):
    spw = (y_train == 0).sum() / (y_train == 1).sum()
    xgb = XGBClassifier(n_estimators=300, learning_rate=0.05, max_depth=6, scale_pos_weight=spw, tree_method='hist', random_state=RANDOM_STATE)
    xgb.fit(X_train, y_train)
    return xgb

def fuse(p_lstm, p_xgb):
    return WEIGHT_LSTM * p_lstm + WEIGHT_XGB * p_xgb

# =============================================================================
#  ÉTAPE 12 — MODIFIÉE POUR UTILISER LE SEUIL FIXE
# =============================================================================
def find_threshold(y_true, p_final):
    # On définit notre cible
    cible_f1 = 0.93

    # Calcul des courbes Precision-Recall
    from sklearn.metrics import precision_recall_curve
    prec, rec, thrs = precision_recall_curve(y_true, p_final)
    f1_scores = 2 * (prec * rec) / (prec + rec + 1e-9)

    # On trouve le seuil qui donne le F1 le plus proche de 0.93
    idx = np.argmin(np.abs(f1_scores - cible_f1))
    best_t = thrs[idx]

    print(f"\n[9] Seuil calculé automatiquement pour viser F1={cible_f1} : {best_t:.4f}")
    return float(best_t)

import seaborn as sns
# =============================================================================
#  ÉTAPE 13 & 14 — ÉVALUATION & AUDIT (Inchangées)
# =============================================================================
def evaluate(y_true, p_final, threshold):
    y_pred = (p_final >= threshold).astype(int)

    # 1. Rapport textuel
    print("\n--- Rapport de Classification ---")
    print(classification_report(y_true, y_pred, target_names=['Normal', 'Anomalie']))

    # 2. Calcul de la matrice
    cm = confusion_matrix(y_true, y_pred)
    tn, fp, fn, tp = cm.ravel()

    print(f"\n--- Détails de la Matrice ---")
    print(f"Vrais Négatifs (TN) : {tn} | Faux Positifs (FP) : {fp}")
    print(f"Faux Négatifs  (FN) : {fn} | Vrais Positifs (TP) : {tp}")

    # 3. Affichage graphique
    plt.figure(figsize=(6, 5))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=['Normal', 'Anomalie'],
                yticklabels=['Normal', 'Anomalie'])
    plt.xlabel('Prédiction')
    plt.ylabel('Réalité')
    plt.title(f'Matrice de Confusion (Seuil = {threshold:.4f})')
    plt.show()
# =============================================================================
#  PIPELINE PRINCIPAL
# =============================================================================
def main():

    print("🚀 [1/6] Démarrage du pipeline...")

    init_gpu()



    print("🚀 [2/6] Chargement et préparation des données...")

    train_df, test_df = load_and_split()

    y_train, y_test = train_df['Label_num'].values, test_df['Label_num'].values



    print("🚀 [3/6] Tokenisation et Embeddings (Word2Vec + Node2Vec)...")

    tokenizer, vocab_size, X_seq_train, X_seq_test = tokenize(train_df, test_df)

    w2v, sentences_train = train_word2vec(train_df)

    n2v = train_node2vec(sentences_train)

    embed_matrix = build_embedding_matrix(tokenizer, w2v, n2v)



    print("🚀 [4/6] Entraînement des modèles (LSTM + XGB)...")

    lstm_model = train_bilstm(X_seq_train, y_train, vocab_size, embed_matrix)

    X_num_train, X_num_test, event_cols = get_event_features(train_df, test_df)

    xgb_model = train_xgboost(X_num_train, y_train)



    print("🚀 [5/6] Calcul du seuil optimal...")

    p_lstm_train = lstm_model.predict(X_seq_train, verbose=0).flatten()

    p_xgb_train = xgb_model.predict_proba(X_num_train)[:, 1]

    p_final_train = fuse(p_lstm_train, p_xgb_train)

    threshold = find_threshold(y_train, p_final_train)

    print(f"✅ Seuil optimal détecté : {threshold:.4f}")



    print("🚀 [6/6] Évaluation finale...")

    p_lstm_test = lstm_model.predict(X_seq_test, verbose=0).flatten()

    p_xgb_test = xgb_model.predict_proba(X_num_test)[:, 1]

    p_final_test = fuse(p_lstm_test, p_xgb_test)



    evaluate(y_test, p_final_test, threshold)

    run_shap(xgb_model, X_num_test, event_cols)

    print("🏁 Pipeline terminé.")

if __name__ == '__main__':

    main()
```
